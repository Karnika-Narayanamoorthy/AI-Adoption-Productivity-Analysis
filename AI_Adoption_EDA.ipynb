# B. Data understanding and preprocessing

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Deep Learning
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

# Preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Evaluation
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Set visualization style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)

# Create output folder
os.makedirs("results", exist_ok=True)

# Load dataset
df = pd.read_csv("ai_adoption_productivity_2021_2026.csv")

# Display dataset
print("First Five Records:")
display(df.head())

print("\nDataset Shape:", df.shape)
print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())
```

# C. Feature engineering

```python
print(df.columns.tolist())
```

```python
# Remove unwanted spaces from column names
df.columns = df.columns.str.strip()

# Convert YearMonth into Date
df["Date"] = pd.to_datetime(
    df["YearMonth"],
    errors="coerce"
)

# Check conversion
print(df[["YearMonth", "Date"]].head())

# Check missing dates
print("Missing dates:", df["Date"].isnull().sum())

# Sort dataset
df = df.sort_values(
    ["Industry", "Date"]
).reset_index(drop=True)
```

```python
print(df.head())
print(df.columns.tolist())
```

```python
# Extract date-related features
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Quarter"] = df["Date"].dt.quarter
df["Half_Year"] = df["Date"].dt.quarter.apply(
    lambda x: 1 if x <= 2 else 2
)

# Time trend
df["Time_Index"] = (
    (df["Date"].dt.year - df["Date"].dt.year.min()) * 12
    + df["Date"].dt.month
)

# Seasonal encoding
df["Month_Sin"] = np.sin(
    2 * np.pi * df["Month"] / 12
)

df["Month_Cos"] = np.cos(
    2 * np.pi * df["Month"] / 12
)

# Log transformations
df["Log_Active_Users"] = np.log1p(
    df["Global Active Users (Millions)"]
)

df["Log_Tokens"] = np.log1p(
    df["Average Tokens/User/Day"]
)

# Interaction feature
df["Users_Tokens_Interaction"] = (
    df["Global Active Users (Millions)"]
    * df["Average Tokens/User/Day"]
)

# Historical features calculated separately for each industry
group = df.groupby("Industry", sort=False)

df["Previous_Productivity"] = (
    group["Productivity Gain (%)"].shift(1)
)

df["Previous_Users"] = (
    group["Global Active Users (Millions)"].shift(1)
)

df["Previous_Tokens"] = (
    group["Average Tokens/User/Day"].shift(1)
)

df["Productivity_Rolling_Mean"] = (
    df.groupby("Industry")["Productivity Gain (%)"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=3, min_periods=1
        ).mean()
    )
)

df["Users_Rolling_Mean"] = (
    df.groupby("Industry")["Global Active Users (Millions)"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=3, min_periods=1
        ).mean()
    )
)

df["Tokens_Rolling_Mean"] = (
    df.groupby("Industry")["Average Tokens/User/Day"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=3, min_periods=1
        ).mean()
    )
)

# Display engineered dataset
print("Updated Dataset Shape:", df.shape)

display(df.head())

print("\nTotal Available Features:", len(df.columns))
```

# E. Outlier detection using IQR

```python
# Identify outliers using IQR
outlier_columns = [
    "Global Active Users (Millions)",
    "Average Tokens/User/Day",
    "Productivity Gain (%)"
]

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print("\nFeature:", col)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outlier Count:", len(outliers))

# Visualize outliers
plt.figure(figsize=(10, 5))

sns.boxplot(data=df[outlier_columns])

plt.title("Outlier Detection Using Boxplots")
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig("results/outlier_detection.png")
plt.show()
```

# 4. Deep Learning model – MLP

# A. Prepare features and target

```python
# Target variable
target = "Productivity Gain (%)"

# Features used for model training
feature_columns = [
    "Industry",
    "Primary Use Case",
    "Global Active Users (Millions)",
    "Average Tokens/User/Day",
    "Year",
    "Month",
    "Quarter",
    "Half_Year",
    "Time_Index",
    "Month_Sin",
    "Month_Cos",
    "Log_Active_Users",
    "Log_Tokens",
    "Users_Tokens_Interaction",
    "Previous_Productivity",
    "Previous_Users",
    "Previous_Tokens",
    "Productivity_Rolling_Mean",
    "Users_Rolling_Mean",
    "Tokens_Rolling_Mean"
]

X = df[feature_columns].copy()
y = df[target].copy()

# Chronological train-validation-test split
unique_dates = np.sort(df["Date"].unique())

train_end = unique_dates[
    int(len(unique_dates) * 0.70) - 1
]

val_end = unique_dates[
    int(len(unique_dates) * 0.85) - 1
]

train_mask = df["Date"] <= train_end

val_mask = (
    (df["Date"] > train_end) &
    (df["Date"] <= val_end)
)

test_mask = df["Date"] > val_end

X_train, y_train = X[train_mask], y[train_mask]
X_val, y_val = X[val_mask], y[val_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Testing samples:", len(X_test))

print("Training period ends:", train_end)
print("Validation period ends:", val_end)
```

# B. Encode categorical variables and scale numerical variables

```python
# Separate categorical and numerical features
categorical_features = [
    "Industry",
    "Primary Use Case"
]

numerical_features = [
    col for col in feature_columns
    if col not in categorical_features
]

# Numerical preprocessing
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

# Combine transformations
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

# Ensure correct data type
X_train_processed = np.asarray(
    X_train_processed, dtype=np.float32
)

X_val_processed = np.asarray(
    X_val_processed, dtype=np.float32
)

X_test_processed = np.asarray(
    X_test_processed, dtype=np.float32
)

print("Processed training shape:", X_train_processed.shape)
print("Processed validation shape:", X_val_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

# C. Create MLP model

```python
# Build neural network
model = Sequential([
    Input(shape=(X_train_processed.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.2),

    Dense(32, activation="relu"),
    Dropout(0.2),

    Dense(16, activation="relu"),

    Dense(1, activation="linear")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="mean_squared_error",
    metrics=["mae"]
)

# Display architecture
model.summary()
```

# D. Train the model

```python
# Stop training when validation loss stops improving
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

# Train model
history = model.fit(
    X_train_processed,
    y_train,
    validation_data=(
        X_val_processed,
        y_val
    ),
    epochs=150,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)
```

# E. Model evaluation

```python
# Predict test results
y_pred = model.predict(
    X_test_processed
).flatten()

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION RESULTS")
print("------------------------")

print("MAE :", round(mae, 4))
print("MSE :", round(mse, 4))
print("RMSE:", round(rmse, 4))
print("R2 Score:", round(r2, 4))

# Save actual and predicted values
results = pd.DataFrame({
    "Actual Productivity": y_test.values,
    "Predicted Productivity": y_pred
})

results.to_csv(
    "results/prediction_results.csv",
    index=False
)

display(results.head(10))
```
